```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
data=load_breast_cancer()
```

```python
df=pd.DataFrame(data.data,columns=data.feature_names)
df['target']=data.target
df.head()
```

```python
df.isnull().sum()
```

```python
df.loc[4:15,"mean radius"]=None
df.loc[4:15,"mean texture"]=None
```

```python
df.isnull().sum()
```

```python
from sklearn.impute import SimpleImputer
si=SimpleImputer(strategy='mean')
df[['mean radius','mean texture']]=si.fit_transform(df[['mean radius','mean texture']])
```

```python
df
```

```python
import numpy as np
for col in df.select_dtypes(include = np.number):
    min = 3 * -(df[col].std())
    max = 3 * df[col].std()
    df[(df[col] > min) & (df[col] < max)]

df
```
