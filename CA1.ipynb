```python
import pandas as pd
import seaborn as sns
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split

df = sns.load_dataset("tips")

df = df.drop_duplicates()
df = df.dropna()

Q1 = df['total_bill'].quantile(0.25)
Q3 = df['total_bill'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
df = df[(df['total_bill'] >= lower) & (df['total_bill'] <= upper)]

np.random.seed(42)
df['gender'] = np.random.choice(['Male', 'Female'], size=len(df))

def bill_category(bill):
    if bill < 15:
        return "Low"
    elif bill < 25:
        return "Medium"
    else:
        return "High"

df['bill_paid_category'] = df['total_bill'].apply(bill_category)

le = LabelEncoder()
df_label = df.copy()

for col in ['sex', 'smoker', 'day', 'time', 'gender', 'bill_paid_category']:
    df_label[col] = le.fit_transform(df_label[col])

X_label = df_label.drop('bill_paid_category', axis=1)
y_label = df_label['bill_paid_category']

df_onehot = pd.get_dummies(df, drop_first=True)

target_col = [col for col in df_onehot.columns if 'bill_paid_category_' in col][0]

X = df_onehot.drop(target_col, axis=1)
y = df_onehot[target_col]

X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.75, random_state=42)

scaler = StandardScaler()
X_train[['total_bill', 'tip']] = scaler.fit_transform(X_train[['total_bill', 'tip']])
X_test[['total_bill', 'tip']] = scaler.transform(X_test[['total_bill', 'tip']])

print(X_train.shape)
print(X_test.shape)
print(X_train.head())
```
