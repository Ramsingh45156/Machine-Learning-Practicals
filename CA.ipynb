```python
import pandas as pd
import seaborn as sns
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split

df = sns.load_dataset("tips")
df
```

```python
df = df.drop_duplicates()
df = df.dropna()
```

```python
np.random.seed(42)
df['gender'] = np.random.choice(['Male', 'Female'], size=len(df))
```

```python
def bill_category(bill):
    if bill < 15:
        return "Low"
    elif 15 <= bill < 25:
        return "Medium"
    else:
        return "High"

df['bill_paid_category'] = df['total_bill'].apply(bill_category)

print("Dataset Preview:\n", df.head())
```

```python
le = LabelEncoder()

df['sex'] = le.fit_transform(df['sex'])
df['smoker'] = le.fit_transform(df['smoker'])
df['day'] = le.fit_transform(df['day'])
df['time'] = le.fit_transform(df['time'])
df['gender'] = le.fit_transform(df['gender'])
df['bill_paid_category'] = le.fit_transform(df['bill_paid_category'])

print("\nAfter Label Encoding:\n", df.head())
```

```python
X = df.drop('bill_paid_category', axis=1)   
y = df['bill_paid_category']                

print("\nFeatures (X) shape:", X.shape)
print("Target (y) shape:", y.shape)
```

```python
df2 = sns.load_dataset("tips")
df2 = df2.drop_duplicates().dropna()

df2['gender'] = np.random.choice(['Male', 'Female'], size=len(df2))
df2['bill_paid_category'] = df2['total_bill'].apply(bill_category)
```

```python
df2_encoded = pd.get_dummies(df2, drop_first=True)
```

```python

X_onehot = df2_encoded.drop('bill_paid_category_Medium', axis=1)
y_onehot = df2_encoded['bill_paid_category_Medium']

print("\nOne Hot Encoded Data Preview:\n", df2_encoded.head())
print("\nOne Hot Features Shape:", X_onehot.shape)
print("One Hot Target Shape:", y_onehot.shape)
```
