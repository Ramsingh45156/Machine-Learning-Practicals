```python
import pandas as pd
import seaborn as sns
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

df = sns.load_dataset("tips").drop_duplicates().dropna()

np.random.seed(42)
df['gender'] = np.random.choice(['Male', 'Female'], size=len(df))

def bill_category(bill):
    if bill < 15:
        return "Low"
    elif bill < 25:
        return "Medium"
    else:
        return "High"

df['bill_paid_category'] = df['total_bill'].apply(bill_category)

df = pd.get_dummies(df, drop_first=True)

X = df.drop('bill_paid_category_Medium', axis=1)
y = df['bill_paid_category_Medium']

X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.75, random_state=42)

scaler = StandardScaler()
X_train[['total_bill', 'tip']] = scaler.fit_transform(X_train[['total_bill', 'tip']])
X_test[['total_bill', 'tip']] = scaler.transform(X_test[['total_bill', 'tip']])

print(X_train.shape)
print(X_test.shape)
print(X_train.head())
```

```python
import pandas as pd
import seaborn as sns
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split

df = sns.load_dataset("tips")

df = df.drop_duplicates()
df = df.dropna()

Q1 = df['total_bill'].quantile(0.25)
Q3 = df['total_bill'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
df = df[(df['total_bill'] >= lower) & (df['total_bill'] <= upper)]

np.random.seed(42)
df['gender'] = np.random.choice(['Male', 'Female'], size=len(df))

def bill_category(bill):
    if bill < 15:
        return "Low"
    elif 15 <= bill < 25:
        return "Medium"
    else:
        return "High"

df['bill_paid_category'] = df['total_bill'].apply(bill_category)

le = LabelEncoder()
df['sex'] = le.fit_transform(df['sex'])
df['smoker'] = le.fit_transform(df['smoker'])
df['day'] = le.fit_transform(df['day'])
df['time'] = le.fit_transform(df['time'])
df['gender'] = le.fit_transform(df['gender'])
df['bill_paid_category'] = le.fit_transform(df['bill_paid_category'])

X = df.drop('bill_paid_category', axis=1)
y = df['bill_paid_category']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, train_size=0.75, random_state=42
)

scaler = StandardScaler()
X_train[['total_bill', 'tip']] = scaler.fit_transform(
    X_train[['total_bill', 'tip']]
)
X_test[['total_bill', 'tip']] = scaler.transform(
    X_test[['total_bill', 'tip']]
)

print(X_train.shape)
print(X_test.shape)
print(X_train.head())
```
